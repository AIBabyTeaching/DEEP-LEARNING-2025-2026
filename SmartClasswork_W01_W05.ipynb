````python
#
#   UNLOCK — Smart Classwork W01-W05 — Eng. Ahmed Metwalli
#   Enter your registration number and the passcode below.
#   Do NOT modify this cell.
#
from IPython.display import display, HTML
import base64, hashlib, json, zlib, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks, regularizers
from sklearn.datasets import make_regression
from IPython.display import display, Markdown, HTML

warnings.filterwarnings("ignore")

# ── Encrypted tokens (do not modify) ─────────────────────────────────────────
_VERIFY_TOKEN    = "NjUT+sx84fVVBaWtHxfe9SpmnPDsDlBUTSa8pzWqOQ=="
_QUESTION_PAYLOAD = "NjX1l5eDNQrUdflBLFYq7KyjFeM7RpsE3nnmh5YBMHbyWKXcodTmzixM8pz2YOFwxZZkfJnipjIuzckNVZJptvzgeGtbl1Q5+WbUYiAewzNowGfnIv+kTnNXaE9ykgUWwUH0P25eeRw/WE21wU/d0gR7u7oRTvua1Vr7xk7DF94UzuogiOxhIVanIK9eqyzhbSl/cXGbNZ+Z6a4ybl4P41l+ndHDOQ3Osm+bFlSTv5BfcwNxuq5l3iDad9gQ0ntEjAMLC+yHrEGiMAK3kx+B+M374eYqmUZnTspOD+h7WavzJOxwtEgD9C50lnPPWlUmlUwMgGjAK5Gk3bVfDaPiNlEm2hAnzqjPYrzdfz5JE7Dy4548x0/qr36IY7kYpW/TECU8e3+tRhvKNCXjsCnshiASQKd7yH7Bu1wZr0RzXCOuNx3jnHGiogJ+CohDhwoSpJ145DMO5m8KzCmQ+tcIDUpf+szBi7HkluQ34XFBRBatTM/GX27CjDBILwwXp/xbadXYHtXMbtdAK+etfJ0PhxQmYoWlGFWTyH7wKpvS2dn1SuAbdzJOWtC1u0KzKcJsT9Cp7+wUHI6+8P8CQKTyyWdmiolkFLKaoqBswCdgNukJNNzLGm8gCx9bSFissJ/1LApz9XDu2kywmcc39unjDdQvSxdJMrC/CbUW8vYkcdvvhe0PJGurQIvMagTwsh3j4W9uOKpiDJwouYn0lSK5N+J6KAPEWE6DEx+CTGz5nGOYtvbM5w8jmySI7HtbQg7bGNJyNBA/jUZStg8yIePXmrabMESMxqpFqKlSSkhiDZLJUXf1jB0q9575tz7DgEhDU3Yt7BmgZQAAc5Ro7iTo7Vkzpm3KTS8d/BXCdNm1ui6k3N1TTIMIOIyHQE1ON9iUowbvgEfbqb32IovO7zGBKOo/3K6e678zsckE0HzJgISSE84LwOlZ+DGSZW52NwUvPsjXUuVAOIaRHhvvq124jgS2ZMckiW97laWoT5SGQ6XrxsoF959i/7llDWwPk+XpXMFjNDebHFU7NMKBLFir2FKv8CqVqZutilhVrbTujxdiaLDaFgZ1REUDvX9s66U+xGPS0njPadMukIxALoV0IPdkkkJHKJIp0slfrJBWyhrDCYaHD6ItMitnawlpIZ6AUIUQIkTU0tfVCUUlgiraJnIQU4FBONXnGnQzaZssWuxskQEKe8QTLew6F1hQ+rr3+svMqFQxu7bsy4wUerUl66R17PcBR0DTYfcDPhZL1ast2kE1ws/tTFGH0cJp2xVEE/XlZXZEn98cu8MVfMlLmRi58Mm9YoD89Q7zt57zog7P7rkBJM4+Z6tX7I4eJbrqsh9Sj/ndcZALDm972PwLCTbIsXREk4lH0EidZbUVsligBBqia3w9zakOWdq6NwX63dn94OyM+zER4IGhxzGVx9fATawhSowJyoElmiKi4wZ716oT4F7TS8EA757ZGOnISsEhTeHtHwRlqWh73sccLBJtBJcyfejX8JMbslksEQpk6sOGOYEJTm5mFfTgFd6nIEk4ZORK8zCZpG4Jv6/ApDM57NdSneOZWgCDgIucDGm1ufyNLJAh4OoGLMkd+EjzPP+HHtiBAF9sunFJah7r/XUU9r1LYLCf48GoyA7XGLLydh9csf3fmHmYWfJo2ewCZYbF9ac+3narSDRStbH5nPpJUzfsMwDKR46lBSdHwb+PBMtcCnYMdYkW1IKcO8Fx3B6h4sS2eidE4MgMcYqTodHqLQjhSP2MB/tiaZupEALvuV+GBAeWKEftvUsPFJgld7JqcGUXUVG4+/3RoLdxNnirzMnJvUAwHoO/LgHytDp0ol0hdyGeqfRO9VnwRIEc6nCpS8XIpD5cvKzGQBJaIpRTTCudRtHwEidxLtwF542Fm1ouE265OBX7u0A8znAbBZACDdN6I00LdMvAdzaMN2Lqb9MasL71bD97vivH4A/a+bPZ0k6109gKdLt2CfSiXZK18zJ2s1TBnS8GY1IU5Kvc8+U11gi6eXvtKJ7ToNmSpvYikFV3mEJdBstDCnnh6ztNed1IteAMUHFDjv/jjLd8dosrJnta8TL7cwXrohRU0bI1rdpulkIlwd6+2cu3lqZGF5QCoC7wsPU9nz1H58C3ewUfN2ihY3zRGhRfwrhE5lGkpm7al0mEvDQLMenvJeBY3Eg/dxIqexdWt8OjwFaX45QYJ7MH5TkhhuASSkTMrwE8t2KcFgnbXQoU3zR7Y5AtVTVFVmGPkgfNwVeu4VOKgHVPkzKXx321nO7cW0D5+xbm54BoSJQwoq6V9dxBimpaqhmtDTpkdX+EsbRypMfkKjVRH9zVxulSi5sbVdIXieNm+ggjC1Re+0pW8hT8EsQwq9kGuVVS9Wgy0qt3nzva2lEEmV6ne+HISWVQrXfP1SEULimrtmsHZ9DORjGA5CP+QDPztaED8VhkbIX6AjQrju8Qa1WowYV8+fSTKR1hjO4f85syDFNmeMzgNL27F0FFYvH3Q9cj5BfSfZfy250ltk28i0dhT6L77tr1DvRn5UmpyVS8NotYFUC48WiKwlqNRgKxKBMBGcu+NxtgIxANKnTp1yy25uV/8n/5bQ0nnGo6rSYHb3LOeM5jytgo6xbQ1PHiZ5cTo9yRhu1CHyvoVpDtLNxGBC7fXqxVLfQndX9x+7dKtAJ3/O0ULZPH8jHzgmgwxoOhqRhMySiB/BSn6k0yMdpBPojK3QK6v/llNjVIA34seyQiV+sw+eBqIjTXJMUve4Wg4yFwJztNEjmsJONB/LdcYkCTKRISMJonrPMAN8Cm91nhn9MM6Z3PqXHUhh1VIjZC/PNgQ/AV+6mtbx3U8vQuWv+ZA/mGdCK5hl07S/GasJ28RpYmUc556l2FaVng3Qtl8ljr9Wg0NUKYvreMxcgtiyLSfQXDrcgBpiMymzz8b6IJ+LF6A3862u3+ukv6XsUvEGSKksvb1I53mUfQG3XDjKkcl4xGlXLEJPWZwokaPKyraKZ/eWMsirqzmpBmcYno0CGcNFBpf4Wn/idDUaWZEDbf4Byohh7MWh2HoY+6uIlulrjAW0dRuP++UYTZ2eir0rxJtBxF1oktL/LZvscCyJ/SLaUiKz9UlY6XiByz/a+L4x1PUJLEbEiokemvbyGZNtJoMXHmF1IyNtNUQvLEXWBKZcN3Gifg2cnSWafoGPyEqrwQg0MtgVZmw6rlx/Rv33yoXhZdQ55RVbBQhUmuVfwFRvFk1lWBB82hchsAbWD9UDQNYOSO7H7rE3qwvfT3GHUwE8bH2MwjX3BKo0TDV0OFT6mSeGHveWFZQ0WvMTg4+qSTCVXJBfy2s5sdp73ddHbhgwjVmKmCSxl8H2DJYL9Vi9j7VrWkfbdkuEVeIMGwi48NyK1/UD0ooOnJfbjviBXS+GFH8GX/EwJyAhnOXfWG8a05h+IECCfdcvQCN0JCNRWrjBtBRhyTsfFDiZPmUNkv+zPuWqQcmStF+vvXJC6JkoRmJgjmeLTwi7UlVpwvDEyyRJgWWFJv0qFFkLilPQUtBrGoFrUZiATYLQBM4/HXB63KGmIhDbv6DN/c85FVUbsiSsDoKF+Zwg5g0b/EfO+dW8fcPIGizwe5UC2kCY5NiM5HKtYQlqZCK1ErI47UR68t6nXKtKqqUDdNXKFg5FGelIJ+2IRKYec49xUacp4YVH8gmxqMGBU9fGFSLUB8zDotpkQYZwMeO+BDNQ/5j3cuarB6kVyLCqa/lQkjyY1NJcpNTRS4DQzMKsHciGepI4KBCwae8w658iTK21KEaxq4jBEg4G5UtPwmWejlGB3/isSRSoEdvCI+QUpgbd3aLdYhJ5xVihKHZo0oItfV7RkOWLXR65mX7nIdIOL562/LYgU86njIcH66w0uk7p8MbBi58VnN6zTvdptwHAvoGDsZhQ55HSpn45bUl/kJDbAPH04TJg4K51nJ3uu4Lsd6vHRkU3L/79ZPRB5LRizIXbTH8O6XH0AeQ8Lflw7mu26pzFFShG0zq/0+1GOgAy+YsolNMwTCUbbU5aEVFV+o6uDf5ZrZzi3CctOlqsvQpd2oS9ztQAGp+l7/BupK2kIyifdzRk9kEz+zUMQ55h3i5gvdGtw8z+YTff9agl6g8bMxssJr49pfLXd+A/MDyec7UoJy2PUkE13n2Vs5tsR64Zz9LMg8SZvVvqegbS34grbMr8BIigTph3NHVKRRtl+kl85RbZhT5I1xABDO2e5KgkSLgpSAwmfnfmYfrsJKJcn2xHGwoBsPvuuqNhBPFsBtQkJE7i8r3JH+vH7RX2hmESzTeROwWaj3k5qFY+ZbFLDVqb+FI5lYLPpTlhQ0tDxgARyqDgdE6APNunscFYGIfvvjPSN7iUGB2iAvr4JBZDElqcxa4HnYxLXuoDDEOKCb4bPar5N6z9EL0Qzc6nqi2bbXnMT0zEknjWWyxFAHXOPrUz1LWEf51E5UX+kbDxgSlARbkVBnrS2iDlwuqaNSEqTnS4k+YCEK/C32gsdCzMCIQhMIWzIum5+22LqbNSEr3FKJYHscptGyHds5jvSrgUGtTAWqYaWTJoPXcPVoHpNgyViAP5q5hn+znanR11Vjk8AVr9HIdbX5NnK1da+2Se/GD5bEO7UOX5mNGQjex/1yP6jNsFSk8OOaR1fvDubj57rDfoOId0mYbVPYPJ8uLFHcXnGxfSqP1XaFm91DNUBcIWvhh/PyVfwJ4t2KlSWmttNMVa8yc4eSMVP8c6CHf8WyyflsaY26GdmNJUtdma/QPeQXczjkJlVcch/1NgjxAUuIBbNLW6HHquajKlGRx0mlPlB3DD/eVgl9A58eT/VZA6p7n0i1obO2Gj2vNLPTa1C2064AWLYpYTfyvFZqCVEkv9vf2dWFNB9LIL5/wMvKZu2dJlsrtgxZfgX7jp5khB5RDo/c9JW9FK8KkMDlH+LOaf8Lwqo8QO7toasXVvreBhr4YpGn9QMEBfjZHabAo5vYSboXNhPSmavINXL0J4Q+jIssJXpr9SHWx+YW5VtG81gtRIvy/+khngQDpY79dOjC3856eSjLpbWW3Ekuq/glHtzrZt6k2eJvFPO7/CEwKQ2NfvZ2ulBcgHwTLXHHu87eeCwuy2vLapisTRt5wUPfN95MxMeoACCXS9JenxJ66hEhjY8Rz2doyxoYEJEkdyXVoJbx1r9E4Bz/P/WAiA=="

# ── Crypto helpers ────────────────────────────────────────────────────────────
def _xor(data, pwd):
    key = hashlib.sha256(pwd.encode()).digest()
    return bytes([b ^ key[i % len(key)] for i, b in enumerate(data)])

def _unlock(encoded, pwd):
    try:
        return zlib.decompress(_xor(base64.b64decode(encoded), pwd)).decode("utf-8")
    except Exception:
        return None

# ── Classwork data loader ─────────────────────────────────────────────────────
def _make_classwork_data(seed):
    # Fashion MNIST
    (x_tr, y_tr), (x_te, y_te) = keras.datasets.fashion_mnist.load_data()
    x_tr = x_tr.astype("float32") / 255.0
    x_te = x_te.astype("float32") / 255.0
    x_tr = x_tr[..., np.newaxis]   # (60000,28,28,1)
    x_te = x_te[..., np.newaxis]   # (10000,28,28,1)
    x_val, y_val = x_tr[-10000:], y_tr[-10000:]
    x_tr,  y_tr  = x_tr[:-10000],  y_tr[:-10000]

    # Optional regression toy-data (W01 continuity)
    rng = np.random.default_rng(seed)
    x_reg, y_reg = make_regression(n_samples=140, n_features=1,
                                    noise=float(rng.integers(8, 20)),
                                    random_state=seed % (2**31))
    x_reg = x_reg.reshape(-1)
    order = np.argsort(x_reg)
    x_reg, y_reg = x_reg[order], y_reg[order]

    return x_tr, x_val, x_te, y_tr, y_val, y_te, x_reg, y_reg

# ── Prompt ────────────────────────────────────────────────────────────────────
print("=" * 60)
print("    SMART CLASSWORK UNLOCK    W01-W05")
print("=" * 60)

_reg = input("\n  Enter your 9-digit registration number: ").strip()
if not (_reg.isdigit() and len(_reg) == 9):
    display(HTML("<div style='color:#e94560;font-weight:bold;font-size:15px'>"
                 "Invalid registration number — must be exactly 9 digits.<br>"
                 "Re-run this cell and try again.</div>"))
    raise ValueError("Invalid registration number")

_key = input("  Enter the classwork passcode: ").strip()

if _unlock(_VERIFY_TOKEN, _key) != "W01_W05_DL_CLASSWORK_OK":
    display(HTML("<div style='color:#e94560;font-weight:bold;font-size:15px'>"
                 "Incorrect passcode — ask your instructor and re-run.</div>"))
    raise ValueError("Wrong passcode")

_QUESTIONS = json.loads(_unlock(_QUESTION_PAYLOAD, _key))

_seed = int(hashlib.sha256(f"{_key}|{_reg}".encode()).hexdigest(), 16) % (2**32)
X_train_cnn, X_val_cnn, X_test_cnn, \
y_train_cnn, y_val_cnn, y_test_cnn, \
x_reg, y_reg = _make_classwork_data(_seed)

_fp = hashlib.sha256(f"{_key}|{_reg}".encode()).hexdigest()[:12].upper()

CLASS_NAMES = ["T-shirt/top","Trouser","Pullover","Dress","Coat",
               "Sandal","Shirt","Sneaker","Bag","Ankle boot"]

display(HTML(f"""
<div style='background:#0d2818;color:#c9d1d9;border-radius:12px;padding:22px;
            font-family:monospace;border:1px solid #238636;'>
  <h3 style='color:#3fb950;margin-top:0'>&#x1F513; Classwork Unlocked Successfully</h3>
  <table style='width:100%;border-collapse:collapse'>
    <tr><td style='color:#8b949e;padding:3px 10px'>Registration</td>
        <td style='color:#fff;padding:3px 10px'><strong>{_reg}</strong></td></tr>
    <tr><td style='color:#8b949e;padding:3px 10px'>Fingerprint</td>
        <td style='color:#fff;padding:3px 10px'><strong>{_fp}</strong></td></tr>
    <tr><td style='color:#8b949e;padding:3px 10px'>CNN data</td>
        <td style='color:#fff;padding:3px 10px'>
          <strong>{X_train_cnn.shape[0]:,} train · {X_val_cnn.shape[0]:,} val · {X_test_cnn.shape[0]:,} test
          — shape {X_train_cnn.shape[1:]}</strong></td></tr>
    <tr><td style='color:#8b949e;padding:3px 10px'>Scope</td>
        <td style='color:#fff;padding:3px 10px'>
          <strong>W01_intro_dl · W02_perceptron_mlp · W03_optimization
          · W04_cnn_basics · W05_cnn_advanced_aug_transfer</strong></td></tr>
  </table>
</div>
"""))

# ═══════════════════════════════════════════════════════════════════════════════
# PROVIDED HELPER FUNCTIONS — read these before solving tasks!
# ═══════════════════════════════════════════════════════════════════════════════

def plot_history(history, title="Training History"):
    """Plot accuracy and loss curves for a Keras history object."""
    h = history.history if hasattr(history, "history") else history
    fig, axes = plt.subplots(1, 2, figsize=(13, 4))
    for ax, metric in zip(axes, ["accuracy", "loss"]):
        ax.plot(h.get(metric, []), label=f"train")
        ax.plot(h.get(f"val_{metric}", []), label="val", linestyle="--")
        ax.set_title(f"{title} — {metric}")
        ax.set_xlabel("Epoch")
        ax.set_ylabel(metric)
        ax.legend()
        ax.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()


def show_sample_images(X, y, class_names, n=10):
    """Display n sample images with their class labels."""
    plt.figure(figsize=(15, 3))
    for i in range(min(n, len(X))):
        plt.subplot(1, n, i + 1)
        plt.imshow(X[i].squeeze(), cmap="gray")
        plt.title(class_names[y[i]], fontsize=8)
        plt.axis("off")
    plt.tight_layout()
    plt.show()


def show_augmented_grid(images, augmenter, n_aug=3):
    """Show original images alongside augmented versions."""
    n = min(5, len(images))
    fig, axes = plt.subplots(n, n_aug + 1, figsize=(3 * (n_aug + 1), 3 * n))
    for i in range(n):
        img = tf.expand_dims(images[i], axis=0)
        axes[i, 0].imshow(images[i].squeeze(), cmap="gray")
        axes[i, 0].set_title("Original", fontsize=8)
        axes[i, 0].axis("off")
        for j in range(n_aug):
            aug = augmenter(img, training=True)[0].numpy()
            axes[i, j + 1].imshow(aug.squeeze(), cmap="gray")
            axes[i, j + 1].set_title(f"Aug {j+1}", fontsize=8)
            axes[i, j + 1].axis("off")
    plt.suptitle("Original vs Augmented", fontsize=11)
    plt.tight_layout()
    plt.show()


def preprocess_for_mobilenet(X, target_size=(96, 96)):
    """Resize grayscale Fashion MNIST images to target_size and convert to RGB."""
    X_resized = tf.image.resize(X, target_size)
    X_rgb     = tf.image.grayscale_to_rgb(X_resized)
    return X_rgb.numpy().astype("float32")


def build_transfer_model(base_model, num_classes=10, dropout_rate=0.3):
    """Add a classification head (GAP→Dense→Dropout→softmax) on top of base_model."""
    inputs  = keras.Input(shape=base_model.input_shape[1:])
    x       = base_model(inputs, training=False)
    x       = layers.GlobalAveragePooling2D()(x)
    x       = layers.Dense(64, activation="relu")(x)
    x       = layers.Dropout(dropout_rate)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)
    return keras.Model(inputs, outputs, name="transfer_model")


def plot_multi_history(histories, labels):
    """Overlay validation accuracy and loss curves for multiple experiments."""
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    for h_obj, label in zip(histories, labels):
        hd = h_obj.history if hasattr(h_obj, "history") else h_obj
        axes[0].plot(hd.get("val_accuracy", []), label=label)
        axes[1].plot(hd.get("val_loss",     []), label=label)
    axes[0].set_title("Validation Accuracy")
    axes[1].set_title("Validation Loss")
    for ax in axes:
        ax.set_xlabel("Epoch")
        ax.legend()
        ax.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()


def display_results_table(results_table):
    """Display a list-of-dicts as a formatted pandas DataFrame."""
    df = pd.DataFrame(results_table)
    if not df.empty:
        display(df)
    return df


def compare_results(acc1, acc2,
                    label1="Task 4 (Feature Extraction)",
                    label2="Task 5 (Fine-Tuned)"):
    """Print a side-by-side accuracy comparison."""
    print(f"{'-'*52}")
    print(f"  {label1:<30}: {acc1*100:.2f}%")
    print(f"  {label2:<30}: {acc2*100:.2f}%")
    diff = (acc2 - acc1) * 100
    sign = "+" if diff >= 0 else ""
    print(f"  Improvement : {sign}{diff:.2f}%")
    print(f"{'-'*52}")


# Pre-compute MobileNet inputs (runs once, ~30 s on CPU)
print("\nPre-processing Fashion MNIST for MobileNetV2 (96×96 RGB) …")
print("  (this may take ~30 s on CPU — please wait)")
X_train_mob = preprocess_for_mobilenet(X_train_cnn)
X_val_mob   = preprocess_for_mobilenet(X_val_cnn)
X_test_mob  = preprocess_for_mobilenet(X_test_cnn)
print(f"  Done. X_train_mob shape: {X_train_mob.shape}")

print("\n✅ All helper functions and data are ready. Start with Task 1!")
````
